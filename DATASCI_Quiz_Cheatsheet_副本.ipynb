## Quick keyword map

| If the question says... | Use... |
|---|---|
| read in / import file | `pd.read_csv(...)` |
| assign to variable | `x = ...` |
| descriptive statistics | `.describe()` |
| 3rd element | `[2]` |

| concatenate | `+` |
| double indexing | `[i][j]` |
| 75 evenly spaced points | `np.linspace(start, stop, 75)` |
| array | `np.array([...])` |
| mean | `np.mean(...)` |
| sum | `np.sum(...)` |
| ln | `np.log(...)` |
| e to a power | `np.exp(...)` |
| square root | `np.sqrt(...)` |
| sin / cos | `np.sin(...)`, `np.cos(...)` |
| pi | `np.pi` |
| scatter plot | `plt.scatter(x, y)` |
| line plot | `plt.plot(x, y)` |
| bar plot | `plt.bar(...)` |
| is equal | `==` |
| not equal | `!=` |
| contains / membership | `in` |
| both | `and` |
| at least one | `or` |
| opposite | `not` |
| exactly one | `^` |
| approximately equal floats | `np.isclose(x, y)` |
| if / otherwise | `if ... else` |
| multiple cases | `if ... elif ... else` |
| swap values | temporary variable |

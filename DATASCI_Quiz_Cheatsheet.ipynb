# 14. Copying lists

This:

```python

```

does **not** make an independent copy of a list.

Both variables point to the same list object.

Use:

```python
b = a.copy()
```

to make a separate list.


````python
a = [1, 2, 3]
b = a

a[0] = 100
print("a:", a)
print("b:", b)

c = a.copy()
a[1] = 200

print("a after another change:", a)
print("independent copy c:", c)
````

# 38. Function composition `f(g(t))`

If

\[
f(x)=x^2-5x+2
\]

and

\[
g(t)=\sin(t+3),
\]

compute `g(t)` first, then substitute it everywhere `x` appears in `f`.


````python
t = np.linspace(2*np.pi, 4*np.pi, 75)

g = np.sin(t + 3)
f_of_g = g**2 - 5*g + 2

plt.plot(t, f_of_g)
plt.xlabel("t")
plt.ylabel("f(g(t))")
plt.title("Function composition")
plt.show()
````

# 40. Floating-point equality

Do not rely on exact `==` for floating-point numbers.

Safer approaches:

```python
abs(x - y) < tolerance
```

or

```python
np.isclose(x, y)
```


````python
x = 0.1 + 0.1 + 0.1
y = 0.3

print(x == y)

tol = 1e-12
print(abs(x - y) < tol)

print(np.isclose(x, y))
````

# 47. `if / elif / else`

Once one condition is True, the later branches are skipped.


````python
score = 85

if score >= 90:
    print("A")
elif score >= 80:
    print("B")
else:
    print("Below B")
````

# 48. Interval conditions

Python allows chained comparisons:

```python

5 <= n <= 10
10 < n < 15
```


````python
n = 8

if 1 <= n < 5:
    print("1 <= n < 5")
elif 5 <= n <= 10:
    print("5 <= n <= 10")
elif 10 < n < 15:
    print("10 < n < 15")
else:
    print("n >= 15")
````

# 51. Password-checking logic

If the program should report only the **first** failed requirement, use one `if / elif / elif / else` chain rather than several independent `if` statements.


````python
proposed_password = "Peter123"

if len(proposed_password) < 8:
    print("Password is too short")
elif proposed_password.isalpha():
    print("Password needs a non-alphabetic character")
elif proposed_password.islower():
    print("Password needs an uppercase letter")
else:
    print("Password set successfully")
````

# 52. Swapping two variables

This does **not** work:

```python

a = b
```

because the original value of `b` is lost.

Use a temporary variable:


````python
a = "left"
b = "right"

temp = a
a = b
b = temp

print("a =", a)
print("b =", b)
````

# 54. Threshold-testing pattern

A common pattern is:

1. compute a value
2. compare it with a threshold
3. use `if / else`


````python
c = np.array([1, 2, 3])
total = np.sum(c)

if total >= 5:
    print("The sum is greater than or equal to 5")
else:
    print("It is strictly less than 5")
````
